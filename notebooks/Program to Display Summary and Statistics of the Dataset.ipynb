```python
# Load the libraries
import numpy as np
import pandas as pd

# Open the CSV file
df = pd.read_csv("iot_synthetic.csv")

# Display the dataset
df
```

## Feature Information

| Feature      | Meaning / Description                   | Type        |
|--------------|-----------------------------------------|-------------|
| timestamp    | Date and time of record                 | Datetime    |
| device       | IoT device type (doorbell, webcam, etc.)| Categorical |
| packet_rate  | Number of packets in 5 seconds          | Numeric     |
| byte_rate    | Number of bytes in 5 seconds            | Numeric     |
| conn_count   | Number of active connections            | Numeric     |
| tcp_syn      | Number of SYN packets                   | Numeric     |
| udp_ratio    | Fraction of UDP packets (0–1)           | Numeric     |
| label        | Class of record (benign or attack)      | Categorical |
